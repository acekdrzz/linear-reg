```python
"""
our value class that we borrowed from main linear regression repo
with impoved backward function which supports bigger datasets


"""

class Value:
    def __init__(self, data, children=(), _op='', label=''):
        self.data = data
        self.grad = 0.0
        self._backward = lambda: None
        self._prev = set(children)
        self._op = _op
        self.label = label

    def __repr__(self):
        return f"Value(data={self.data})"

    def __add__(self, other):
        other = other if isinstance(other, Value) else Value(other)
        out = Value(self.data + other.data, (self, other), '+')
        
        def _backward():
            self.grad += 1.0 * out.grad
            other.grad += 1.0 * out.grad
        out._backward = _backward
        return out
    def __truediv__(self, other):
        other = other if isinstance(other, Value) else Value(other)
        out = Value(self.data / other.data, (self, other), '/')
        def _backward():
            self.grad += (1.0 / other.data) * out.grad
            other.grad += (-self.data / other.data**2) * out.grad
        out._backward = _backward
        return out

    def __pow__(self, other):
        assert isinstance(other, (int, float)), "only int/float powers"
        out = Value(self.data**other, (self,), f'**{other}')
        def _backward():
            self.grad += other * (self.data**(other - 1)) * out.grad
        out._backward = _backward
        return out
            

    def __mul__(self, other):
        other = other if isinstance(other, Value) else Value(other)
        out = Value(self.data * other.data, (self, other), '*')
        def _backward():
            self.grad += other.data * out.grad
            other.grad += self.data * out.grad
        out._backward = _backward
        return out
    def __rmul__(self,other):
        return self * other
    def __sub__(self, other):
        other = other if isinstance(other, Value) else Value(other)
        out = Value(self.data - other.data, (self, other), '-')
        def _backward():
            self.grad += 1.0 * out.grad
            other.grad += -1.0 * out.grad
        out._backward = _backward
        return out
        

    def backward(self):
        topo = []
        visited = set()
        stack = [(self, iter(self._prev))]
        visited.add(self)
        while stack:
            node, children = stack[-1]
            found_child = False
            for child in children:
                if child not in visited:
                    visited.add(child)
                    stack.append((child, iter(child._prev)))
                    found_child = True
                    break
            if not found_child:
                topo.append(node)
                stack.pop()
        self.grad = 1.0
        for node in reversed(topo):
            node._backward()
```

```python
"""
part of code that i use to extract the information from csv to python as a tupple of column headers
and keys that is equipoise to following row

"""

filename  = "Food_Delivery_Times(numerical only).csv"
with open(filename, "r")as csvfile:
    csvreader = csv.DictReader(csvfile)
    dataset = []
    for row in csvreader:
        dataset.append(row)
TrainingSet = [
    [Value(float(v)) if v.strip() != '' else Value(0.0) for v in list(row.values())[:3]]
    for row in dataset
]
CheckSet = [Value(float(row_values[3])) for row_values in (list(row.values()) for row in dataset)]
```

```python
"""
part that we implement our weights and biases for the first time
for now i implemented weight and biases just for 4 variables since we will start by
only numerical features first 
"""
w = [Value(random.uniform(0,5)) for i in range(3)]
b = Value(random.uniform(0,5))
```

```python
"""
here is our prediction function which multiplies the weights with inputs and adds
biases to our wx this part also in charge of handling empty strings at our csv
"""


def pred():
    predictions = []
    for row in TrainingSet:          # row = one row's list of feature Values, e.g. [dist, prep, exp]
        y = sum((wi * xi for wi, xi in zip(w, row)), b)
        predictions.append(y)
    return predictions
```

```python
"""

here we will implement our learning rate (which in this case stored in variable #alpha#) and 
number of iteration (in other words epoch) besides our loss function which we will use to 
evaluate the convergence of our alghoritm

"""
iteration = 1000
alpha = 0.0000000001
y = pred()
loss = sum(((xi - yi_true) ** 2 for xi, yi_true in zip(y, CheckSet)), Value(0))
```

```python
"""
at this part we will start our gradient decsent and iterate trough it at the value we gave
to iteration and w learning rate alpha 


"""

for i in range(iteration):
    y = pred()
    loss = sum(((xi - yi_true) ** 2 for xi, yi_true in zip(y, CheckSet)), Value(0))
    for x  in w:
        x.grad  = 0
    b.grad = 0
    loss.backward()
    for weight in w:
        weight.data -= weight.grad * alpha
    b.data-= b.grad * alpha
    print(loss,end="\r")
```
